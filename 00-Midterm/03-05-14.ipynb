**Elapsed Time**
Write a Python program that accepts a **start time** and a **finish time**, both occurring on the same day, and calculates the elapsed time between them.

Implement the following functions:

read_hour() – returns the hour in 24-hour format (valid values: 0–23).
read_minute() – returns the minute (valid values: 0–59).
read_second() – returns the second (valid values: 0–59).
to_seconds(h,m,s) – returns the total number of seconds represented by h hours, m minutes, and s seconds.
elapsed_time(start,finish) – returns the elapsed time, in seconds, from start to finish.
**Hint:**
For example, if the start time is 08:15:00 and the finish time is 13:20:00, you can calculate the elapsed time by first converting both times into the total number of seconds since 00:00:00, and then subtracting the start time (in seconds) from the finish time (in seconds).

**Example**
Start time:
>> Enter hour: 8↵
>> Enter minute: 15↵
>> Enter second: 0↵
Finish time:
>> Enter hour: 13↵
>> Enter minute: 20↵
>> Enter second: 0↵
Elapsed time is 5 hours 5 minutes 0 seconds.

```python
total_seconds=0
hour=0
minute=0
second=0
def to_seconds(h,m,s):
    total_seconds=s+(m*60)+(h*3600)
    return total_seconds 
def read_hour():
    hour=int(input("Enter hour: "))
    return hour
def read_minute():
    minute=int(input("Enter minute: "))
    return minute
def read_second():
    second=int(input("Enter second: "))
    return second
## get a total_sec
def elapsed_time(start_time,finish_time):
    total_seconds=abs(start_time-finish_time)
    
    return f"{total_seconds//3600} hours {(total_seconds%3600)//60} minutes {(total_seconds%3600)%60} seconds."


# *****************************************

def read_time():
    print('>> ', end='')
    hour = read_hour()
    print('>> ', end='')
    minute = read_minute()
    print('>> ', end='')
    second = read_second()
    return to_seconds(hour, minute, second)

print('Start time:')
start_time = read_time()
print('Finish time:')
finish_time = read_time()
print('Elapsed time is', elapsed_time(start_time, finish_time))
```
