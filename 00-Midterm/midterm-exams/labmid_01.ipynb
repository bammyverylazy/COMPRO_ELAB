**SECRET TEXT**

Write a Python program that compares a string `guess` againt a target string `secret` (both having the same length) and checks each character position based on the following rules:
- **`B` (Bulls / Exact match):** The character matches the exact position in `secret`.
- **`C` (Cows / Partial match):** The character exists in `secret`, but at a different position.
- **`N` (None):** The character does not exist in `secret` at all.

**Example**
| Secret | Guess | Expected Output | Explanation |
| :--- | :--- | :--- | :--- |
| `APPLE` | `ALBUM` | `B C N N N` | `A` (Index 0 exact match = B), `L` (Exists elsewhere = C), `B, U, M` (Not found = N) |
| `TRAIN` | `TRACK` | `B B B N N` | `T, R, A` (Index 0, 1, 2 exact match = B), `C, K` (Not found = N) |

```python
secret = input("secret: ")
guess = input("guess : ")

ans = ''
for i in range(len(secret)):
    if guess[i] == secret[i]:
        ans+='B'
    elif guess[i] in secret:
        ans+='C'
    else:
        ans+='N'
        
print(f"Result: {ans}")
```
