Consider the given BankNote and Coin classes as follows.

Complete the Python program that uses these classes to exchange a given amount of money (in baht) into banknotes and coins. The available denominations are:

Banknotes: 1,000, 500, 100, 50, and 20 baht
Coins: 10, 5, 2, and 1 baht
The program should report the number of each banknote and coin used for the exchange.

The program must exchange the money using the largest denomination available before using smaller denominations. For example, when exchanging 60 baht, the program should give one 50-baht banknote and one 10-baht coin instead of three 20-baht banknotes.

```python
# given
class Coin:
    def __init__(self, value = 1):
        self.value = value

    def __str__(self):
        return f'{self.value} Baht Coin'

class BankNote:
    def __init__(self, value = 20):
        self.value = value

    def __str__(self):
        return f'{self.value} Baht Banknote'
```

```python
user = int(input())
# if user>= 1000:
#     print(f"You get {user//1000} of {BankNote(1000)}")
#     user = user%1000
bank = [1000,500,100,50,20]
coin = [10,5,2,1]

for value in bank:
    if user//value >0 :
        print(f"You get {user//value} of {BankNote(value)}")
        user = user%value

for value in coin:
    if user//value >0 :
        print(f"You get {user//value} of {Coin(value)}")
        user = user%value
```
