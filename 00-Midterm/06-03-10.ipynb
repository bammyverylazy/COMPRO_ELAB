**Count Vowels and Consonants**
Write a Python program that reads a string and identifies all distinct vowels and distinct consonants appearing in it.

A vowel is one of the letters a, e, i, o, u. Any other English alphabet letter is considered a consonant.

The program should:

ignore letter case,
ignore non-alphabetic characters,
output all vowels and consonants in lowercase, and
display each group in alphabetical order.
Hint: Try the str.lower() and str.upper() methods.

Example 1
Enter a string: Apple↵
Vowels: a e 
Consonants: l p 
Example 2
Enter a string: Apple Banana Carrot↵
Vowels: a e o 
Consonants: b c l n p r t 

```python
string = input("Enter a string: ").strip().lower()

vowels = "aeiou"
vowel_set = []
consonant_set = []

for char in string:
    if char in vowels and char not in vowel_set:
        vowel_set.append(char)
    elif char.isalpha() and char not in consonant_set and char not in vowel_set:
        consonant_set.append(char)
vowel_set.sort()
consonant_set.sort()
print("Vowels: ",end="")
print(" ".join(vowel_set))
print("Consonants: ", end="")
print(" ".join(consonant_set))
```
