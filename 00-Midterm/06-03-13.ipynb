**Robot Simulator**
A robot starts at coordinate (0,0) on a 2D grid and is initially facing North (N). The robot receives a sequence of \(n\) commands that determine how it moves or changes direction.

Each command is represented by a single character:

F — Move forward 1 unit in the direction the robot is currently facing.
L — Turn left 90 degrees without changing its position.
R — Turn right 90 degrees without changing its position.
The robot executes the commands one by one in the given order.

Write a Python program that simulates the robot's movement and prints:

The robot's final x-coordinate and y-coordinate.
The robot's final facing direction (N, E, S, or W).
Example 1
Commands: FFRFL↵
Final position: (1,2)
Facing: N
Example 2
Commands: FFLFFRFRFF↵
Final position: (0,3)
Facing: E

```python
'''
1. start at (0,0) N
input = sequence of n commands
    F +1
    L left90
    R right90

'''
position = [0,0]
directions = ['N','E','S','W']
facing = 'N'
i =0
commands = input("Commands: ")

for c in commands:
    direct_index = directions.index(facing)
    if c == 'R':
        facing = directions[(direct_index+1)%4]
    elif c == 'L':
        facing = directions[(direct_index-1)%4]
    else:
        if facing == 'N':
            position[1]+=1
        elif facing == 'S':
            position[1]-=1
        elif facing == 'W':
            position[0]-=1
        elif facing == 'E':
            position[0]+=1
    #print(f"Command: {c} | Movement: {position}, {facing}")
print(f"Final position: ({position[0]},{position[1]})")
print(f"Facing: {facing}")

```
