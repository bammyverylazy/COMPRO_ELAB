```python
# given
print(f"o=============================o")
print(f"      ALARM CONFIGURATION      ")
print(f"o=============================o")
entry = input("Set alarm (HH:MM): ")
hour, minute = map(int, entry.split(":"))
alarm = SnoozableAlarm(hour, minute)
basic = Alarm(hour, minute)
alarm.set_snooze_minutes(int(input("Snooze every (minutes): ")))
print(f"o=============================o")
print(f"          ALARM BEGIN          ")
print(f"o=============================o")
print("Actions:")
print("   1: Change snooze duration")
print("   2: Set alarm time")
print("   3: Start alarm")
print("   4: Stop alarm")
print("   5: Snooze")
print("   6: Rewind alarm time")
print("   7: Compare both alarm types")
print("   0: Exit")
print(f"[{alarm.get_time()}] ALARM RINGING! ALARM RINGING!")

action = input(f"[{alarm.get_time()}] Action: ")
while True:
    if action == "0":
        print("Alarm program ended.")
        break
    if action == "1":
        new_minutes = input(
            f"[{alarm.get_time()}] New snooze duration (minutes): "
        )
        alarm.set_snooze_minutes(int(new_minutes))
        print(
            f"[{alarm.get_time()}] Snooze duration changed to "
            f"{alarm.get_snooze_minutes()} minutes."
        )
    elif action == "2":
        new_time = input(f"[{alarm.get_time()}] New alarm time (HH:MM): ")
        hour, minute = map(int, new_time.split(":"))
        alarm.set_time(hour, minute)
        basic.set_time(hour, minute)
        print(f"[{alarm.get_time()}] Alarm time changed.")
    elif action == "3":
        alarm.start()
        print(f"[{alarm.get_time()}] ALARM RINGING! ALARM RINGING!")
    elif action == "4":
        result = alarm.respond("STOP")
        if result == "OFF":
            print(f"[{alarm.get_time()}] Alarm stopped.")
        else:
            print(f"[{alarm.get_time()}] Alarm is already off.")
    elif action == "5":
        result = alarm.respond("SNOOZE")
        if result == "SNOOZED":
            print(
                f"[{alarm.get_time()}] Snoozed by "
                f"{alarm.get_snooze_minutes()} minutes "
                f"({alarm.get_snooze_count()} so far)."
            )
        else:
            print(
                f"[{alarm.get_time()}] "
                "Alarm is off. Start it before snoozing."
            )
    elif action == "6":
        amount = input(f"[{alarm.get_time()}] Rewind by (minutes): ")
        alarm.subtract_minutes(int(amount))
        print(f"[{alarm.get_time()}] Alarm time moved back.")
    elif action == "7":
        print(f"[{alarm.get_time()}] Same respond('SNOOZE') call on both:")
        for clock in [basic, alarm]:
            answer = clock.respond("SNOOZE")
            print(
                f"[{alarm.get_time()}]   {type(clock).__name__} "
                f"-> {answer}, now {clock.get_time()}"
            )
    else:
        print(f"[{alarm.get_time()}] Invalid action. Please choose 0 to 7.")
    action = input(f"[{alarm.get_time()}] Action: ")
```
