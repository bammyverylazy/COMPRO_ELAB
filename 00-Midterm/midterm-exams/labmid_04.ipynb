**Bomb Distance Map Grid**

Write a Python program that accepts the grid size `n` and the bomb coordinates (`rows`, `cols`), then prints an `n` x `n` grid where each cell displays its Manhattan distance to the bomb location (`|i - rows| + |j - cols|`), based on the following rules:

- The cell at the bomb coordinates `(rows, cols)` displays `0`.
- Every other cell `(i, j)` displays its calculated distance from the bomb, separated by spaces.

**Example**

| Input | Expected Output | Explanation |
| :--- | :--- | :--- |
| <pre>5<br>2<br>2</pre> | <pre>4 3 2 3 4 <br>3 2 1 2 3 <br>2 1 0 1 2 <br>3 2 1 2 3 <br>4 3 2 3 4 </pre> | Creates a 5x5 grid with the bomb located at index (2, 2). Distance increases outward based on Manhattan distance. |
| <pre>4<br>0<br>0</pre> | <pre>0 1 2 3 <br>1 2 3 4 <br>2 3 4 5 <br>3 4 5 6 </pre> | Creates a 4x4 grid with the bomb located at the top-left corner (0, 0). |

```python
n = int(input("Enter n: "))
rows = int(input("Enter r: "))
cols = int(input("Enter c: "))

for i in range(n):
    t = ''
    for j in range(n):
        if i==rows and j==cols:
            t+='0'
        else :
            distance = abs(i-rows)+abs(j-cols)
            d = str(distance)
            t+=f'{d}'
        
        t+=' '
    print(t) 
```
