```python
class Room:
    hotel_name = "ABC Hotel" # Class attributes: Room.hotel_name, Room.total_rooms
    total_rooms = 0
    def __init__(self,room_number,room_type,price_per_night):
        self.room_number = room_number
        self.room_type = room_type
        self.price_per_night = price_per_night
        self.is_available = True
        Room.total_rooms +=1
    def book(self):
        if self.is_available:
            self.is_available = False
            print(f"Room {self.room_number} has been booked.")
        else:
            print("Error: Room is not available.")
    def checkout(self):
        if not self.is_available:
            self.is_available = True
            print(f"Room {self.room_number} is now available.")
        else:
            print("Error: Room is not currently booked.")
    def calculate_price(self,nights):
        return self.room_number*nights
    def display_room(self):
        print(f"Room Number: {self.room_number}")
        print(f"Room Type: {self.room_type}")
        print(f"Price per Night: {self.price_per_night:.2f} Baht")
        print(f"Status: {"Available" if self.is_available else "Not Available"}")
    @classmethod
    def set_hotel_name(cls,name):
        cls.hotel_name = name
    @classmethod
    def get_total_rooms(cls):
        return cls.total_rooms
    @staticmethod
    def is_valid_nights(nights):
        return nights > 0
    @staticmethod
    def calculate_discount(nights):
        if nights >=1 and nights<=2:
            return 0
        elif nights<=4:
            return 0.5
        else:
            return 0.10
```

```python
# Main program
room1 = Room(101, "Standard", 1500)
room2 = Room(201, "Deluxe", 2500)
room3 = Room(301, "Suite", 4000)

print(f"Hotel: {Room.hotel_name}")
print()

room1.display_room()
print()

room2.display_room()
print()

room3.display_room()
print()

print("Booking Room 201...")
room2.book()
print()

print("Booking Room 201 again...")
room2.book()
print()

nights = 5

if Room.is_valid_nights(nights):
    subtotal = room2.calculate_price(nights)
    discount_rate = Room.calculate_discount(nights)
    discount = subtotal * discount_rate
    total = subtotal - discount

    print(f"Number of nights: {nights}")
    print(f"Price per Night: {room2.price_per_night:.2f} Baht")
    print(f"Subtotal: {subtotal:.2f} Baht")
    print(f"Discount: {discount_rate:.0%}")
    print(f"Total Price: {total:.2f} Baht")
else:
    print("Error: Invalid number of nights.")

print()

print("Checking out Room 201...")
room2.checkout()
print()

print("Checking out Room 101...")
room1.checkout()
print()

Room.set_hotel_name("XYZ Hotel")
print(f"Hotel name changed to {Room.hotel_name}.")
print()

print(f"Total rooms: {Room.get_total_rooms()}")
```
