**Digit Places**
Write a Python program that accepts an integer num (where num is between 0 to 9,999) and implements the following functions to return the digit at each place value:

ones(num) – returns the ones (units) digit of num.
tens(num) – returns the tens digit of num.
hundreds(num) – returns the hundreds digit of num.
thousands(num) – returns the thousands digit of num.
In addition, implement a function sum_digits(num) that returns the sum of all the digits of num.
**Example**
Enter a number (0 to 9999): 8143↵
Ones place is 3.
Tens place is 4.
Hundreds place is 1.
Thousands place is 8.
Sum of all digits is 16.

```python
def ones(num):
    return num % 10

def tens(num):
    return (num // 10) % 10

def hundreds(num):
    return (num // 100) % 10

def thousands(num):
    return (num // 1000) % 10

def sum_digits(num):
    return ones(num) + tens(num) + hundreds(num) + thousands(num)

num = int(input("Enter a number (0 to 9999): "))

print(f"Ones place is {ones(num)}.")
print(f"Tens place is {tens(num)}.")
print(f"Hundreds place is {hundreds(num)}.")
print(f"Thousands place is {thousands(num)}.")
print(f"Sum of all digits is {sum_digits(num)}.")
```
