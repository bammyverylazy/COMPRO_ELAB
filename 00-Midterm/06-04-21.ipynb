**Image Shift**
Write a Python program that shifts an image to the right horizontally and downward vertically by the number of positions specified by the user. Any empty positions created after the shift should be filled with the digit '0'.

The program first asks the user to enter the horizontal shift size and the vertical shift size. It then accepts an input image of any rectangular size, entered one line at a time. The user enters -1 to indicate the end of the image input.

After the image has been shifted, the program displays the resulting image.

Enter vertical shift size: 2↵

```python
shift_row = int(input("Enter horizontal shift size: "))
shift_col = int(input("Enter vertical shift size: "))

print("Enter image:")
image = []

while True:
    user = input()

    if user == "-1":
        break

    image.append(list(user))


height = len(image)
width = len(image[0])

shifted_image = []

for i in range(height):
    shifted_image.append(['0'] * width)


for i in range(height):
    for j in range(width):
        new_row = i + shift_col
        new_col = j + shift_row

        if new_row < height and new_col < width:
            shifted_image[new_row][new_col] = image[i][j]

print("After shift:")

for row in shifted_image:
    print("".join(row))
```
