**Box with an X Pattern**

Write a Python program that accepts an integer `n` and prints an `n` x `n` grid forming a box with a diagonal cross (X inside a box).

**Example**

| Input | Expected Output | Explanation |
| :--- | :--- | :--- |
| `7` | <pre>.......<br>..   ..<br>. . . .<br>.  .  .<br>. . . .<br>..   ..<br>.......</pre> | Draws a 7x7 box with borders and diagonal cross lines with a perfect single center point at row index 3. |

```python
n = int(input("Enter n: "))

for i in range(n):
    for j in range(n):
        if i==0 or j==0 or i==n-1 or j==n-1:
            print(".",end='')
        elif i==j:
            print(".",end='')
        elif i+j==n-1:
            print(".",end='')
        else :
            print(" ",end='')
    print()
```
