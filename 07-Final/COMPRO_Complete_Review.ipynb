### Why `[1,2,3] + 3` fails

List `+` means **concatenate another iterable/list**.  //**To concatenate means to link or join separate things together in a series or chain.**

```python
[1,2,3] + [4]       # works
[1,2,3] + 4         # TypeError
```

But:

```python
[1,2,3] * 3
```

means repeat the list three times.

This is **not scalar multiplication**.

NumPy arrays behave differently:

```python
np.array([1,2,3]) * 3
# [3, 6, 9]
```

# 7. `lambda`, `sorted`, and `key` — understand this deeply

The pattern is:

```python
sorted(data, key=lambda item: ...)
```

The `lambda` tells Python **what value to use for comparison**.

Example:

```python
orders = {
    "A": 500,
    "B": 200,
    "C": 900
}

sorted(orders.items(), key=lambda item: item[1])
```

`orders.items()` produces pairs:

```text
("A", 500)
("B", 200)
("C", 900)
```

So:

```python
lambda item: item[1]
```

means **sort by the second element**, the amount.

Descending:

```python
sorted(orders.items(), key=lambda item: item[1], reverse=True)
```

### Important syntax trap

Wrong:

```python
orders.items
```

Correct:

```python
orders.items()
```

`.items` is the method object. `.items()` calls it.

# 12. NumPy reshape — understand dimensions

If an array has `12` elements, valid reshapes must also contain `12` elements:

```text
3 × 4 = 12
2 × 6 = 12
1 × 12 = 12
```

So:

```python
np.arange(12).reshape(3, 4)
```

works.

But:

```python
np.arange(12).reshape(5, 3)
```

does not, because `5 × 3 = 15`.

### `-1`

NumPy can infer one dimension:

```python
np.arange(12).reshape(3, -1)
# 3 × 4
```

Only one dimension can be inferred.

# 14. NumPy boolean masks — essential for lab problems

Boolean comparison:

```python

```

produces a Boolean array.

Then:

```python

```

selects only matching values.

Examples:


### Important

With NumPy arrays, use:

```python
&   # and
|   # or
~   # not
```

and put each comparison in parentheses.

```python
(a > 5) & (a < 10)
```

Do not use Python's `and` / `or` for element-wise array comparisons.

# 16. NumPy transformations — calibration pattern

A common lab pattern is:

```python
calibrated = original[:k] * scale + offset
```

This means:
1. take only the first `k`
2. multiply element-wise
3. add element-wise

Example:

```python
signals = np.array([10, 20, 30, 40])
k = 2
scale = 3
offset = 5

signals[:k] * scale + offset
# [35, 65]
```

If you need to combine the calibrated prefix with the untouched remainder, think carefully about whether you want:

```python
np.concatenate(...)
```

or an in-place assignment:

```python
result = signals.copy()
result[:k] = result[:k] * scale + offset
```

```python
    # runs regardless
```

### When to use each

**try:** risky operation

**except:** recovery / useful error message

**else:** success path that should only run when the risky operation succeeded

**finally:** cleanup / guaranteed final action

Example:

```python
try:
    number = int(input())
except ValueError:
    print("Invalid integer")
else:
    print(number * 2)
finally:
    print("Finished")
```

# 22. Custom exceptions — advanced exception handling

A custom exception inherits from `Exception`:

```python
class InsufficientBalanceError(Exception):
    pass
```

You can also store information:

```python
class InvalidAgeError(Exception):
    def __init__(self, age):
        self.age = age
        super().__init__(f"Invalid age: {age}")
```

Now:

```python
try:
    check_age(200)
except InvalidAgeError as e:
    print(e)
    print(e.age)
```

### Why `super().__init__()`?

The parent `Exception` class handles the standard exception message.

Your custom class can add structured information while still behaving like a normal exception.

# 25. Class variables vs instance variables

### Instance variable

Different for each object:

```python
class Student:
    count = 0

    def __init__(self, name):
        self.name = name
```

`self.name` belongs to one object.

### Class variable

Shared by the class:

```python
Student.count
```

If incremented through the class:

```python
Student.count += 1
```

it represents shared class-level state.

### Common trap

Do not confuse:

```python
self.count
```

with:

```python
Student.count
```

The first is instance lookup; the second explicitly accesses the class variable.

# 26. `@classmethod` vs `@staticmethod` vs instance method

## Instance method

```python
def method(self):
    ...
```

Needs an object.

## Class method

```python
@classmethod
def method(cls):
    ...
```

Receives the class as `cls`.

Useful for alternative constructors:

```python
@classmethod
def from_string(cls, text):
    name, score = text.split(",")
    return cls(name, int(score))
```

## Static method

```python
@staticmethod
def is_valid_score(score):
    return 0 <= score <= 100
```

No `self` or `cls` is automatically supplied.

### Mental test

Ask:

```text
Does this need object state? → self
Does this need class state / create cls? → cls
Does it need neither? → staticmethod
```

# 29. Abstract Base Classes (ABC)

When a parent class defines an interface that subclasses **must implement**:

```python
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass
```

Then:

```python
class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def area(self):
        return self.w * self.h
```

You cannot instantiate the abstract class:

```python
# Shape()  → TypeError
```

The subclass must implement the abstract method.

# 30. `dataclass`

For classes that mostly store data, `dataclass` removes repetitive boilerplate.

```python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    score: float
```

Python generates useful methods such as:
- `__init__`
- `__repr__`
- equality comparison

You can still define methods:

```python
@dataclass
class Student:
    name: str
    score: float

    def passed(self):
        return self.score >= 50
```

### Frozen dataclass

```python
@dataclass(frozen=True)
class Species:
    name: str
    rarity: str
```

This is useful when an object should be immutable.

# 31. DTO thinking

A DTO (Data Transfer Object) is mainly a structured object for moving data between parts of a program.

Typical pattern:

```python
@dataclass
class PlayerDTO:
    user_id: str
    name: str
    score: int

    def to_dict(self):
        return {
            "user_id": self.user_id,
            "name": self.name,
            "score": self.score
        }
```

And:

```python
@classmethod
def from_dict(cls, data):
    return cls(
        user_id=data["user_id"],
        name=data["name"],
        score=data["score"]
    )
```

This `to_dict` / `from_dict` pattern is worth understanding because it combines:
- dataclasses
- dictionaries
- class methods
- serialization concepts

# 32. Properties

A property lets you expose method-like logic through attribute syntax.

```python
class Student:
    def __init__(self, score):
        self._score = score

    @property
    def score(self):
        return self._score

    @score.setter
    def score(self, value):
        if not 0 <= value <= 100:
            raise ValueError("Score must be 0..100")
        self._score = value
```

Now:

```python
s.score = 90
print(s.score)
```

The setter performs validation.

### Why `_score`?

It is a convention meaning "internal/protected-style attribute"; it is not true private access control.

# 33. Composition vs inheritance

Inheritance:

```text
Dog IS-A Animal
```

Composition:

```text
Car HAS-A Engine
```

Example:

```python
class Engine:
    def start(self):
        return "engine started"

class Car:
    def __init__(self):
        self.engine = Engine()
```

Prefer composition when an object is made from other objects rather than being a specialized form of them.

# 36. `map`, `filter`, comprehensions

## `map`

```python
nums = [1,2,3]
```

Often clearer:

```python
squares = [x*x for x in nums]
```

## Filter

```python
evens = list(filter(lambda x: x % 2 == 0, nums))
```

Equivalent comprehension:

```python
evens = [x for x in nums if x % 2 == 0]
```

Know both styles, but choose the clearer one.
