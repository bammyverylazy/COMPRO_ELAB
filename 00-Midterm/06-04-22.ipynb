**Tic-Tac-Toe Validator**
Tic-Tac-Toe is a two-player game played on a 
3
×
3
 board. Each cell on the board contains one of the following characters:

X — occupied by Player X
O — occupied by Player O
. — an empty cell
A player wins if they have three of their marks in a row, either:

Horizontally
Vertically
Diagonally
Write a Python program that reads the current Tic-Tac-Toe board and determines the game result. You may assume that the input always represents a valid game state, so there will never be a board where both players win or where the number of moves is invalid.

The possible results are:

X wins — Player X has three marks in a row.
O wins — Player O has three marks in a row.
Draw — The board is full and neither player has won.
Game not finished — The board still contains at least one empty cell and neither player has won.
Example 1
The current board:
XOX↵
OXO↵
OXX↵
Result: X wins
Example 2
The current board:
XXX↵
OO.↵
...↵
Result: X wins
Example 3
The current board:
XOX↵
OOX↵
XO.↵
Result: O wins
Example 4
The current board:
XOX↵
OXX↵
OXO↵
Result: Draw
Example 5
The current board:
XO.↵
OX.↵
...↵
Result: Game not finished
Example 6
The current board:
XOX↵
OX.↵
OXO↵
Result: Game not finished
Explanation: The program determines the result based on the current state of the board, not on the eventual outcome of the game.

```python
print("The current board:")

board = []

for i in range(3):
    user = input()
    user = [i for i in user]
    board.append(user)

case = ""
winner = ""
win = False
game_finished = True

# Horizontal
for i in range(3):
    if board[i][0] == board[i][1] == board[i][2] and board[i][0] != ".":
        win = True
        case = "Horizontally"
        winner = board[i][0]
        break

# Vertical
if not win:
    for i in range(3):
        if board[0][i] == board[1][i] == board[2][i] and board[0][i] != ".":
            win = True
            case = "Vertically"
            winner = board[0][i]
            break

# Diagonal
if not win:
    if (board[0][0] == board[1][1] == board[2][2]
            and board[0][0] != "."):
        win = True
        case = "Diagonally"
        winner = board[1][1]

    elif (board[0][2] == board[1][1] == board[2][0]
            and board[0][2] != "."):
        win = True
        case = "Diagonally"
        winner = board[1][1]

# Check whether board is full
for row in board:
    if "." in row:
        game_finished = False
        break

# Result
if win:
    print(f"Result: {winner} wins")
elif not game_finished:
    print("Result: Game not finished")
else:
    print("Result: Draw")
```
