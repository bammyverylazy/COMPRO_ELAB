**Approximate Substring**
A substring is a contiguous sequence of characters within a text.

Given a text and a pattern, your task is to find all occurrences of the pattern in the text using the following rules:

An exact match occurs when every character in the pattern is identical to the corresponding character in the substring.
An approximate match occurs when the substring differs from the pattern in exactly one character position.
Exact matches have higher priority than approximate matches.
If the text contains one or more exact matches, display only the exact matches.
Approximate matches are considered only when no exact match exists in the text.
For an approximate match, replace the different character with a question mark ('?') before displaying it.
Write a Python program that reads a text and a pattern, then displays the text with every matching substring enclosed in square brackets ('[' and ']').

The text is scanned from left to right. Whenever a match is found, the search continues immediately after the matched substring. Therefore, overlapping matches are not allowed.

You may assume that:

the pattern is not longer than the text;
the text and the pattern contain only English letters ('A'–'Z' and 'a'–'z');
uppercase and lowercase letters are considered different characters.
Example 1
Text: acabababababcbabab↵
Pattern: aba↵
ac[aba]b[aba]babcb[aba]b
Example 2
Text: acababababaccbabab↵
Pattern: abc↵
ac[ab?]b[ab?]b[a?c]b[ab?]b
Example 3
Text: AAAGTGTGTCTGAC↵
Pattern: GTAT↵
AAA[GT?T][GT?T]GAC

```python
text = input("Text: ")
pattern = input("Pattern: ")

m = len(pattern)
matches = []

# First, look for exact matches.
i = 0
while i <= len(text) - m:
    if text[i:i + m] == pattern:
        matches.append((i, pattern))
        i += m
    else:
        i += 1

# If there are no exact matches, look for approximate matches.
if not matches:
    i = 0
    while i <= len(text) - m:
        substring = text[i:i + m]
        differences = 0
        match = list(substring)

        for j in range(m):
            if substring[j] != pattern[j]:
                differences += 1
                match[j] = '?'

        if differences == 1:
            matches.append((i, ''.join(match)))
            i += m
        else:
            i += 1

# Build the output.
result = []
last = 0

for start, matched in matches:
    result.append(text[last:start])
    result.append('[' + matched + ']')
    last = start + m

result.append(text[last:])

print(''.join(result))
```
