### A3. Write a word counter

Implement `count_words(words)` using a dictionary (not Counter first). Normalize with `.strip().lower()`. Then write a second version using `collections.Counter`.


````python
def count_words(words):
    counted = {}
    for word in words:
        word = word.strip().lower()
        if word not in counted:
            counted[word] = 0
        counted[word]+=1
    return counted


print(count_words([" Cat ", "dog", "cat", "DOG", "cat"]))
# -> {"cat": 3, "dog": 2}
````

### B4. Matrix multiplication from scratch

Implement `matrix_multiply(A, B)` using ordinary nested lists and three loops. Do not use NumPy `@`, `dot`, or `matmul`. Validate rectangular inputs and inner dimensions; raise `ValueError` when invalid.


````python
# GPT
def matrix_multiply(A, B):
    # Validate that both matrices are non-empty
    if not A or not B:
        raise ValueError("Matrices must not be empty")

    # Validate that each matrix is rectangular
    if not all(isinstance(row, list) for row in A + B):
        raise ValueError("Matrices must contain lists as rows")

    if not all(len(row) == len(A[0]) for row in A):
        raise ValueError("A must be rectangular")

    if not all(len(row) == len(B[0]) for row in B):
        raise ValueError("B must be rectangular")

    # Inner dimensions must match
    if len(A[0]) != len(B):
        raise ValueError("Incompatible matrix dimensions")

    result = []

    for i in range(len(A)):
        row = []
        for j in range(len(B[0])):
            total = 0
            for k in range(len(A[0])):
                total += A[i][k] * B[k][j]
            row.append(total)

        result.append(row)

    return result


A = [[1, 2], [3, 4]]
B = [[5, 6], [7, 8]]

print(matrix_multiply(A, B))
# [[19, 22], [43, 50]]
````

### D3. Instance, class, and static methods


````python
class Temperature:
    def __init__(self, value):
        self.value = value
    def as_fahrenheit(self):
        return self.value * 9 / 5 + 32
    @classmethod
    def from_fahrenheit(cls, value):
        return cls((value - 32) * 5 / 9)
    @staticmethod
    def is_valid(value):
        return isinstance(value, (int, float))

t = Temperature.from_fahrenheit(68)
print(t.value)
print(t.as_fahrenheit())
print(Temperature.is_valid("68"))
````
