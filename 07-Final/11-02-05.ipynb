[Branch Sales] สรุปยอดขายแต่ละสาขา
ร้านค้ามีหลายสาขาและขายสินค้าหลายชนิด ตารางยอดขายหนึ่งแถวแทนหนึ่งสาขา และหนึ่งคอลัมน์แทนสินค้าหนึ่งชนิด

โจทย์ จงสร้างโปรแกรมที่รับตารางยอดขายขนาด R แถว C คอลัมน์ แล้วใช้ NumPy คำนวณ

ยอดรวมของแต่ละสาขา หรือผลรวมตามแถว
ยอดรวมของสินค้าแต่ละชนิด หรือผลรวมตามคอลัมน์
หมายเลขสาขาที่มียอดรวมสูงสุด โดยหมายเลขสาขาเริ่มจาก 1 และหากเสมอกันให้เลือกหมายเลขน้อยที่สุด
ยอดรวมทั้งหมด
ข้อมูลนำเข้า (Input)
บรรทัดที่ 1 ระบุจำนวนเต็มบวก R C

บรรทัดที่ 2 ระบุยอดขายจำนวนเต็มที่ไม่ติดลบจำนวน R * C ค่า เรียงทีละแถว

ข้อมูลส่งออก (Output)
แสดงผลรวมตามแถวและคอลัมน์ในรูปแบบ NumPy array โดยตรง (สมาชิกคั่นด้วยช่องว่าง ไม่มีจุลภาค) ตามด้วยสาขาที่ดีที่สุดและยอดรวมทั้งหมด

Example 1
Sales table size: 3 3↵
Sales values: 10 20 30 5 5 5 20 20 20↵
Branch totals: [60 15 60]
Product totals: [35 45 55]
Best branch: 1
Grand total: 135

```python
import numpy as np

R, C = map(int, input("Sales table size: ").split())
values = np.array(list(map(int, input("Sales values: ").split())))
sale_values = values.reshape(R, C)
print("Branch totals:", np.sum(sale_values, axis=1))
print("Product totals:", np.sum(sale_values, axis=0))
branch_totals = np.sum(sale_values, axis=1)
best_branch = np.argmax(branch_totals) + 1
print("Best branch:", best_branch)
print("Grand total:", np.sum(sale_values))
```
