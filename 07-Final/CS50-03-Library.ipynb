```python
# Random Library
import random
print(f"random from list *random.choice(seq) => {random.choice(["tail","head"])}")
print(f"random from a seq of int => {random.randint(1,10)}\n")

cards = ["Jack","Queen","King"]
print("Before random.shuffle(list) => " ,cards)
random.shuffle(cards)
print("After => ",cards)

```
