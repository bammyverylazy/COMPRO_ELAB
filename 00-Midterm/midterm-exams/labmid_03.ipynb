**Powers of Integers Up to N**

Write a Python program that accepts an integer `n` and prints all perfect squares ($i^2$) that are less than or equal to `n`

**Example**

| Input | Expected Output | Explanation |
| --- | --- | --- |
| `20` | `1 4 9 16` | Perfect squares $\le 20$ are $1^2, 2^2, 3^2, 4^2$. ($5^2 = 25 > 20$, so it stops) |
| `50` | `1 4 9 16 25 36 49` | Perfect squares $\le 50$ up to $7^2=49$. |

```python
n = int(input("Enter n: "))
i = 1
while True:
    if i**2>n:
        break
    print(i**2,end=' ')
    i+=1
```
