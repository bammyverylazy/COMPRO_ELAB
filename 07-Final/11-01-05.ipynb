Body Status Plot
Write a Python program that asks user a CSV file that contains a list of weight-height pairs (in kilograms and centimeters, respectively). The program then computes the BMI (body mass index) for every pair and displays a scatter plot with the folowing details:

The x-axis is used for weights and labeled as "Weight (kg)"
The y-axis is used for heights and labeled as "Height (cm)"
BMI is color-coded into the plot with the "jet" colormap
A color bar is shown with label "BMI"
Grid lines are also shown
Note that the formula for computing BMI is:
**BMI = weight/height**2**
where weight is in kilograms and height is in meters.

Example 1
Suppose the file body-small.txt contains the data from the table above:

74.8,181.5
86.4,184.0
78.4,184.5
62.0,175.0
81.6,184.0
76.6,180.0
83.6,177.8
The output of the program when specifying body-small.txt as input is:

Enter file name: body-small.txt


```python
import matplotlib.pyplot as plt
import numpy as np

filename = input("Enter file name: ")
data = np.loadtxt(filename , delimiter= ",")
# print(data)

def bmi(h,w):
    h = h / 100
    return w/(h**2)

bmi_values = [ bmi((x[1]),x[0]) for x in data]
plt.scatter(data[:,0],data[:,1],c= np.array(bmi_values))
plt.ylabel("Height (cm)")
plt.xlabel("Weight (kg)")
plt.grid(True)
bar = plt.colorbar()  # add a color bar
bar.ax.set_title("BMI")  # add title to the color bar
plt.set_cmap("jet")  # set color mapping scheme to "jet"
plt.show()
```
